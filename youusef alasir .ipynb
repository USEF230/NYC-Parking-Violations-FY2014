```python
top_streets = df['Street Name'].value_counts().head(10).sort_values().to_frame(name="violations")

plt.figure(figsize=(10, 5))
plt.barh(top_streets.index, top_streets["violations"], color="steelblue")
plt.title("Top 10 Streets by Parking Violations")
plt.xlabel("Number of Violations")
plt.ylabel("Street Name")
plt.show()
```

```python
violation_summary = (
    df.groupby('Violation Description', dropna=True)
    .agg(total_violations=('Summons Number', 'count'))
    .sort_values('total_violations', ascending=False)
    .head(8)
    .sort_values('total_violations', ascending=True)
)

plt.figure(figsize=(11, 5.5))
bars = plt.barh(violation_summary.index, violation_summary['total_violations'], color='#2563EB')

labels = [f"{v/1_000_000:.2f}M" if v >= 1_000_000 else f"{v/1_000:.0f}K" for v in violation_summary['total_violations']]
plt.bar_label(bars, labels=labels, padding=5, fontsize=9, fontweight='bold')

plt.title('Most Common Parking Violations', fontsize=14, fontweight='bold')
plt.xlabel('Total Violations', fontsize=11)
plt.ylabel('Violation Type', fontsize=11)
plt.xlim(0, violation_summary['total_violations'].max() * 1.15) # مساحة فاضية يمين للأرقام
plt.grid(axis='x', linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()
```

```python
county_summary = (
    df.groupby('Violation County', dropna=True)
    .agg(total_violations=('Summons Number', 'count'))
    .sort_values('total_violations', ascending=False)
    .head(8)
)

colors = ['#1E3A8A' if i == 0 else '#60A5FA' for i in range(len(county_summary))]

plt.figure(figsize=(10, 5))
bars = plt.bar(county_summary.index, county_summary['total_violations'], color=colors)

labels = [f"{v/1_000_000:.2f}M" if v >= 1_000_000 else f"{v/1_000:.0f}K" for v in county_summary['total_violations']]
plt.bar_label(bars, labels=labels, padding=4, fontsize=9, fontweight='bold')

plt.title('Top Counties by Parking Violations', fontsize=14, fontweight='bold')
plt.xlabel('County Code', fontsize=11)
plt.ylabel('Total Violations', fontsize=11)
plt.ylim(0, county_summary['total_violations'].max() * 1.15)
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

file_path = '/content/Parking_Violations_Issued_-_Fiscal_Year_2014__August_2013___June_2014_.csv'
df = pd.read_csv(file_path, parse_dates=["Issue Date"], low_memory=False)

valid_years = df[(df['Vehicle Year'] >= 1980) & (df['Vehicle Year'] <= 2015)]

plt.figure(figsize=(10, 5))
sns.histplot(
    data=valid_years,
    x='Vehicle Year',
    bins=20,
    kde=True,
    color='#0284C7'
)

plt.title('Distribution of Vehicle Model Years', fontsize=14, fontweight='bold')
plt.xlabel('Vehicle Year', fontsize=11)
plt.ylabel('Number of Vehicles', fontsize=11)
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()
```

```python
df['Month_Num'] = df['Issue Date'].dt.month

valid_for_agg = df[(df['Vehicle Year'] >= 1980) & (df['Vehicle Year'] <= 2015)]

monthly_pair_df = (
    valid_for_agg.groupby('Month_Num')
    .agg(
        total_violations=('Summons Number', 'count'),
        unique_vehicles=('Plate ID', 'nunique'),
        avg_vehicle_year=('Vehicle Year', 'mean')
    )
    .reset_index()
)

g = sns.pairplot(
    monthly_pair_df[['total_violations', 'unique_vehicles', 'avg_vehicle_year']],
    diag_kind='hist',
    plot_kws={'s': 80, 'color': '#0F766E', 'alpha': 0.85},
    diag_kws={'color': '#2563EB', 'edgecolor': 'white'}
)

g.fig.suptitle('Monthly Metrics Pairplot Overview', y=1.02, fontsize=14, fontweight='bold')
plt.show()
```

```python
top5_makes = df['Vehicle Make'].value_counts().head(5).index

box_df = df[
    (df['Vehicle Make'].isin(top5_makes)) &
    (df['Vehicle Year'] >= 1985) &
    (df['Vehicle Year'] <= 2015)
]

plt.figure(figsize=(10, 5))
sns.boxplot(data=box_df, x='Vehicle Make', y='Vehicle Year', palette='Set2')
plt.title('Vehicle Year Distribution Across Top 5 Makes', fontsize=14, fontweight='bold')
plt.xlabel('Vehicle Make', fontsize=11)
plt.ylabel('Vehicle Year', fontsize=11)
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()
```

```python
num_cols = [
    'Violation Code',
    'Violation Precinct',
    'Issuer Precinct',
    'Vehicle Year',
    'Feet From Curb'
]

corr_df = df[num_cols].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(
    corr_df,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    vmin=-1,
    vmax=1
)
plt.title('Correlation Between Key Numeric Columns', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```
